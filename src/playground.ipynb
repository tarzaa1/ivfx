```python
import time
from data_utils import load_dataset
from ivfx import build_ivf, build_ivfpq, build_ivfrabitq
from ivfx import search, recall, rerank
```

```python
base, queries, gt = load_dataset("imagenet")
d = base.shape[1]
K = 1024
k = 10
c = 50
metric = "l2"
nprobe=16
```

```python
index_ivf = build_ivf(d, base, metric, K)
index_ivfpq = build_ivfpq(d, base, metric, K, 64)
index_ivfrabitq = build_ivfrabitq(d, base, metric, K)
```

```python
t0 = time.perf_counter()
D, I = search(index_ivf, queries, k, nprobe)
print(f"search: {time.perf_counter() - t0:.3f} s")
recall(I, gt, k)
```

```python
t0 = time.perf_counter()
D, I = search(index_ivfpq, queries, c, nprobe)
print(f"search: {time.perf_counter() - t0:.3f} s")
print(recall(I, gt, k))
D, I = rerank(base, queries, I, k, metric)
print(recall(I, gt, k))
print(index_ivfpq.code_size)
```

```python
t0 = time.perf_counter()
D, I = search(index_ivfrabitq, queries, c, nprobe)
print(f"search: {time.perf_counter() - t0:.3f} s")
print(recall(I, gt, k))
D, I = rerank(base, queries, I, k, metric)
print(recall(I, gt, k))
print(index_ivfrabitq.code_size)
```
