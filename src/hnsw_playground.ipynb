```python
import time
from data_utils import load_dataset
from hnswx import build_hnsw, build_hnswsq, build_hnswq8
from hnswx import search, recall, code_size, index_size
```

```python
base, queries, gt = load_dataset("arxiv")
d = base.shape[1]
M = 32
ef_construction = 40
ef_search = 64
k = 10
metric = "ip"
```

```python
t0 = time.perf_counter()
index_hnsw = build_hnsw(d, base, metric, M, ef_construction)
print(f"hnsw   build: {time.perf_counter() - t0:.1f} s")
t0 = time.perf_counter()
index_hnswsq = build_hnswsq(d, base, metric, M, ef_construction)
print(f"hnswsq build: {time.perf_counter() - t0:.1f} s")
# t0 = time.perf_counter()
# index_hnswq8 = build_hnswq8(d, base, metric, M, ef_construction)
# print(f"hnswq8 build: {time.perf_counter() - t0:.1f} s")
```

```python
t0 = time.perf_counter()
D, I = search(index_hnsw, queries, k, ef_search)
print(f"search: {time.perf_counter() - t0:.3f} s")
print(recall(I, gt, k))
print(code_size(index_hnsw))
```

```python
t0 = time.perf_counter()
D, I = search(index_hnswsq, queries, k, ef_search)
print(f"search: {time.perf_counter() - t0:.3f} s")
print(recall(I, gt, k))
print(code_size(index_hnswsq))
```

```python
t0 = time.perf_counter()
D, I = search(index_hnswq8, queries, k, ef_search)
print(f"search: {time.perf_counter() - t0:.3f} s")
print(recall(I, gt, k))
print(code_size(index_hnswq8))
```

```python
for name, idx in [("hnsw", index_hnsw), ("hnswsq", index_hnswsq)]:
    s = index_size(idx)
    print(name, {key: f"{v / 2**20:.1f} MiB" for key, v in s.items()})
```
